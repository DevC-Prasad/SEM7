```python

# Define the function
def function(x):
    return (x + 3) ** 2

# Define the derivative
def gradient(x):
    return 2 * (x + 3)

# Initial parameters
x = 2
learning_rate = 0.1
tolerance = 0.0001
max_iterations = 1000

iteration = 0

# Gradient Descent
while abs(gradient(x)) > tolerance and iteration < max_iterations:

    x = x - learning_rate * gradient(x)

    iteration += 1

# Display results
print("Minimum x value:", x)
print("Minimum y value:", function(x))
print("Iterations:", iteration)
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Define the function
def function(x):
    return (x + 3) ** 2

# Define the gradient
def gradient(x):
    return 2 * (x + 3)

# Initial parameters
x = 2
initial_x = x
learning_rate = 0.1
iterations = 100

# Gradient Descent
for i in range(iterations):
    x = x - learning_rate * gradient(x)

# Generate points for plotting
X = np.linspace(-5, 4, 100)
Y = function(X)

# Plot the function
plt.plot(X, Y, label="y = (x + 3)^2")

# Mark starting point
plt.scatter(initial_x, function(initial_x),
            color='red', label="Starting Point")

# Mark minimum point reached
plt.scatter(x, function(x),
            color='green', label="Minimum Point")

# Graph details
plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent Visualization")
plt.legend()
plt.grid()

plt.show()

# Display results
print("Initial x:", initial_x)
print("Minimum x:", f"{x:.10f}")
print("Minimum y:", f"{function(x):.10f}")
```

```python

import numpy as np
import matplotlib.pyplot as plt


def function(x):
    return (x + 3) ** 2
def gradient(x):
    return 2 * (x + 3)

x = 2
learning_rate = 0.1
iterations = 100

x_values = [x]

for i in range(iterations):
    x = x - learning_rate * gradient(x)
    x_values.append(x)

X = np.linspace(-5, 4, 100)
Y = function(X)

plt.plot(X, Y, label="y = (x + 3)^2")

plt.scatter(x_values[:-1],
            function(np.array(x_values[:-1])),
            color='red', label="Iterations")

plt.scatter(x_values[-1],
            function(x_values[-1]),
            color='green', s=100, label="Minimum")


plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent")
plt.legend()
plt.grid()

plt.show()


print("Minimum x:",f"{x:.10f}")
print("Minimum y:",f"{function(x):.18f}")
```
